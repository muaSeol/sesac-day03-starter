## A. 읽는 순서

**실행** 아래 셀은 할 일이 하나도 없을 때 완료율을 계산합니다. 0으로 나누게 되어 오류가 납니다. 실행한 뒤 아래 설명 순서대로 오류 메시지를 읽습니다.

```python
# 실행: 일부러 오류가 나는 셀입니다.
done_count = 0
total_count = 0
rate = done_count / total_count * 100
print("완료율:", rate)
```

오류 메시지는 맨 아래 줄부터 거꾸로 읽습니다.

1. **맨 아래 줄: 오류 이름과 설명.** `ZeroDivisionError: division by zero` 에서 콜론(:) 앞의 `ZeroDivisionError` 가 오류 이름이고, 뒤의 `division by zero`(0으로 나눔)가 설명입니다.
2. **줄 번호.** `Cell In[1], line 4` 처럼 나옵니다. 노트북에서는 파일 이름 대신 셀이 나오고, `In[ ]` 안의 숫자는 셀을 실행한 순서라 실행할 때마다 바뀝니다. 오류가 난 곳은 `line` 뒤의 숫자로, 셀 안에서 몇째 줄인지입니다. 맨 윗줄(`# 실행` 줄)이 1번입니다.
3. **그 줄의 코드.** 화살표 `---->` 가 가리키는 줄입니다. 이 줄에서 `total_count` 가 0이라 나눌 수 없었습니다.

`.py` 파일을 터미널에서 실행했을 때는 2번 자리에 `File "파일 경로", line 4` 처럼 파일 이름과 줄 번호가 나옵니다. 읽는 순서는 같습니다.
